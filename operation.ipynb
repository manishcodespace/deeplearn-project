## Tensors and Basic Operations

In deep learning, data is typically represented as tensors. Tensors are multi-dimensional arrays, similar to NumPy arrays, but they can also be used on GPUs for accelerated computing. Let's demonstrate a basic mathematical operation using PyTorch, a popular deep learning framework.

```python
# First, let's import the PyTorch library
import torch
```

```python
# Create two sample tensors
tensor_a = torch.tensor([[1, 2], [3, 4]])
tensor_b = torch.tensor([[5, 6], [7, 8]])

print("Tensor A:\n", tensor_a)
print("Tensor B:\n", tensor_b)
```

```python
# Perform a basic mathematical operation: addition
tensor_sum = tensor_a + tensor_b

print("\nResult of Tensor A + Tensor B:\n", tensor_sum)
```
